```python
data = [
	"Melipona scutellaris",
	"Scaptotrigona postica",
	"Tetragonisca angustula",
	"Melipona scutellaris",
	"Scaptotrigona postica",
	"Scaptotrigona depilis",
	"Melipona quadrifasciata",
	"Friesiomelita varia",
	"Scaptotrigona postica"
]

speciesCount = {}
gendersCount = {}
index = 1

print(f"Um total de {len(data)} de ninhos foram observados.")
print("Sendo eles:")
for specie in data:
	print(f"{index} - {specie}")
	index += 1

	speciesCount[specie] = speciesCount.get(specie, 0) + 1
	gender = specie.split(' ')[0]
	gendersCount[gender] = gendersCount.get(gender, 0) + 1

print("\nContagem por espécies:")
for specie, count in speciesCount.items():
	print(f"{specie}: {count}")
print("\nContagem por gêneros:")
for gender, count in gendersCount.items():
	print(f"{gender}: {count}")	

```
