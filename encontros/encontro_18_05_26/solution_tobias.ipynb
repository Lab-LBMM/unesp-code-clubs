```python
from itertools import zip_longest

model = 'ATGCTAGCTAGGCTTACGATCGATCGGATCCTAGGCTAGCTTACGATCGATCGGATCGTACGATCGATGCTAGCTAGGCTTACGATCGATCGGATCCTAG'
muted = 'ATGCTAGCTAGGCTTACGATCGATCGGATCCTAGGCTAGCTTACGATTAGATCGGATCGTACGATCGATGCTAGCTAGGCTTACGATCGATCGGATCCTAG'

def printMutation(type, pos, bas1, bas2):
	print(f"{bas1} -> {bas2} na posição {pos} ({type})")

index = 0
mutations = 0
for index, (bas1, bas2) in enumerate(zip_longest(model, muted, fillvalue='')):
	if bas1 != bas2:
		mutations += 1
		if bas1 == '':
			printMutation("Inserção", index, '-', bas2)
			continue
		elif bas2 == '':
			printMutation("Deleção", index, bas1, '-')
			continue

		bases = [bas1, bas2]
		bases.sort()

		if((bases[0] == 'A' and bases[1] == 'G') or (bases[0] == 'C' and bases[1] == 'T')):
			printMutation("Transição", index, bas1, bas2)
		else:
			printMutation("Transversão", index, bas1, bas2)

	index += 1
		
print("Número de mutações: ", mutations)
print("Taxa de mutação: ", mutations/len(model))
```
