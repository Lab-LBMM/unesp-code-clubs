Você precisa realizar uma investigação sobre quais espécies de plantas são encontradas em uma área específica e sua localização com cordenadas.

para isso serão utilizadas funções len() e set() em conjunto com bibliotecas de manipulação gráfica como matplotlib para observar esses dados de forma visual.

O primeiro passo para rodar o script será chamar as bibliotecas: "os" conversa com o sistema operacional, "argparse" permite ler comandos digitados no terminal, "pandas" lida com análise de dados através de tabelas, "matplotlib" desenha a estrutura do gráfico e "seaborn" aperfeiçoa.

```python
#!/usr/bin/env python3

import os
import argparse
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

O segundo passo é adicionar os argumentos: A primeira linha sinaliza ao script que os argumentos correspondem ao que está no terminal, já os blocos de código subsequentes que começam com "parser.add_argument" são responsáveis por criar os argumentos que serão recebidos pelo código, as linhas "type=" sinalizam que tipo de informação estará em cada argumento e as linhas "required" se esses argumentos são obrigatórios ou opcionais.

```python
parser = argparse.ArgumentParser()
	
parser.add_argument(
	"-e", "--especies",
	type=str,
	required=False,
	help="Path to the input file",
)
parser.add_argument(
	"-x", "--cordenadax",
	type=int,
	required=False,
)
parser.add_argument(
	"-y", "--cordenaday",
	type=int,
	required=False,
)
parser.add_argument(
	"-g", "--grafico",
	type=str,
	required=False,
)
parser.add_argument(
	"-l", "--levantamento",
	type=str,
	required=False,
)

args = parser.parse_args()
```

Agora irei adicionar a parte que irá ser responsável por salvar os dados: A primeira linha checa se os argumentos necessários para adicionar um novo dado foram digitados, as três próximas linhas criam um dicionário que irá conter os novos dados, a linha "dataframe" utiliza o pandas para criar um dataframe contendo essas informações, as próximas checam se o arquivo CSV onde esses dados são armazenados já existe, se ele não existir o cria e se existir o edita.

```python
dados_csv = "dados.csv"

if args.especies and args.cordenadax and args.cordenaday:
    novos_dados = {"especie": [args.especies],
    "x": [args.cordenadax],
    "y": [args.cordenaday]}

    dataframe = pd.DataFrame(novos_dados)

    if not os.path.exists(dados_csv):
        dataframe.to_csv(dados_csv, index=False)
    else:
        dataframe.to_csv(dados_csv, mode="a", header=False, index=False)

    print("Dados salvos")
```

Fazer o levantamento das quantidades de cada espécie: A primeira linha checa se o argumento foi chamado, e a segunda se a pasta com os dados existe, a terceira cria uma variável com o pandas a partir da leitura do arquivo contendo os dados, a quarta cria uma variável com o levantamento de quantas vezes cada espécie ocorre no arquivo utilizando a função ".value_counts()" que atribui um valor a cada espécie presente na coluna espécie e conta quantas vezes elas ocorrem, a última linha mostra o resultado com um "print".

```python
if args.levantamento:
    os.path.exists(dados_csv):
    dados_completos = pd.read_csv(dados_csv)
    levantamento = dados_completos["especie"].value_counts()
    print(levantamento)
```

Agora criar o gráfico: A primeira linha checa se o argumento gráfico foi chamado, a segunda e terceira checam a existência do arquivo base e criam uma variável a partir dele, o bloco que vem a seguir é responsável por criar o gráfico e configurá-lo, o "sns.scatterplot" abre o gráfico e o resto do código determina suas características básicas, a última linha "plt.show()" permite visualizar o gráfico de forma imediata após sua criação.

```python
if args.grafico:
    os.path.exists(dados_csv):
    dados_completos = pd.read_csv(dados_csv)

    sns.scatterplot(data=dados_completos, x="x", y="y", hue="especie", style="especie", s=80, alpha=0.80)
    plt.xlim(0,100)
    plt.ylim(0,100)
    plt.grid(True, linestyle=":", alpha=0.6)
    plt.show()
```
