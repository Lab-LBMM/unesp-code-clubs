Em um alinhamento, temos as seguintes sequências :

seq_1 = 'ATGCTTAGCGTACGATCGATCGATCGATCGATCGATCGATCGA-TCG'
seq_2 = 'ATGCTTAGCGTACGATCGATCGTTTCGATCGATCGATCGATCGACTC'

A partir disso, monte scripts que mostrem:
1) Quantas bases são diferentes entre as duas sequências
2) O percent identity entre as duas sequências (proporção de bases idênticas no alinhamento)

```python
seq_1 = 'ATGCTTAGCGTACGATCGATCGATCGATCGATCGATCGATCGA-TCG'
seq_2 = 'ATGCTTAGCGTACGATCGATCGTTTCGATCGATCGATCGATCGACTC'
```

```python
#Criamos um contador de diferenças
dif = 0
```

```python

#zip junta as sequências das duas listas e compara base a base
#se a é diferente de b, vai para o contador de diferenças
for a, b in zip(seq_1, seq_2):
    if a != b:
        dif += 1


total = len(seq_1)
percent_identity = (total - dif) / total * 100

#Round arredonda um núermo para duas casas decimais 
print("Diferentes pares de base:", dif)
print("Percent identity:", round(percent_identity, 2), "%")
```

```python
for a, b in zip(seq_1, seq_2):
    print(a,b)
```

```python
for a in seq_1:
    print(a)
```
