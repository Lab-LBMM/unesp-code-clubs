Em um alinhamento, temos as seguintes sequências :

seq_1 = 'ATGCTTAGCGTACGATCGATCGATCGATCGATCGATCGATCGA-TCG'
seq_2 = 'ATGCTTAGCGTACGATCGATCGTTTCGATCGATCGATCGATCGACTC'

A partir disso, monte scripts que mostrem:
1) Quantas bases são diferentes entre as duas sequências
2) O percent identity entre as duas sequências (proporção de bases idênticas no alinhamento)

```python
seq_1 = 'ATGCTTAGCGTACGATCGATCGATCGATCGATCGATCGATCGA-TCG'
seq_2 = 'ATGCTTAGCGTACGATCGATCGTTTCGATCGATCGATCGATCGACTC'

# Contadores
diferentes = 0
iguais = 0

# Percorre as sequências ao mesmo tempo
for base1, base2 in zip(seq_1, seq_2):

    if base1 == base2:
        iguais += 1
    else:
        diferentes += 1

# Total de posições do alinhamento
total = len(seq_1)

# Percent identity
percent_identity = (iguais / total) * 100

# Resultados
print(f'Número de bases diferentes: {diferentes}')
print(f'Número de bases idênticas: {iguais}')
print(f'Percent identity: {percent_identity:.2f}%')
```
