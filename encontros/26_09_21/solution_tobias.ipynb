```python
avistamentos = ["teiu", "sabiá", "carcará", "teiu", "arara-canindé", "pombo", "grilo", "carcará", "saruê", "tucano", "cachorro"]
```

```python
contagem = {}
for avistamento in avistamentos:
	if not avistamento in contagem:
		contagem[avistamento] = 0
	contagem[avistamento] += 1
```

```python
def escrever(animal):
	return f"{animal.capitalize()}: {contagem[animal]}"

print("Quantidades de avistamentos:")
for animal in contagem.keys():
	print(f"{escrever(animal)}")	

print(f"\nTotal de animais distintos: {len(contagem.keys())}")
print(f"Total de avistamentos: {len(avistamentos)}")
```
