# Resolução parcial de atividade

 * Identificação de posição das sequências com alteração e retorno das bases alteradas e categoria (transição ou transversão).


```python
sequence1 = 'ATGCTAGCTAGGCTTACGATCGGTCGGATCCTAGGCTAGCTTACGATTAGATCGGATCGTACGATCGATGCTAGCTAGGCTTACGATCGATCGGATCCTAG'
sequence2 = 'ATGCTAGCTAGGCTTACGATCGATCGGATCCTAGGCTAGCTTACGATTAGATCGGATCGTACGTTCGATCCTAGCTAGGCTTACGATCGATCGGATCCTAG'
```

```python
type(sequence1)
```

```python
transicoes = {
    ('A', 'G'), ('G', 'A'),
    ('C', 'T'), ('T', 'C')
}

transversoes = {
    ('A', 'C'), ('C', 'A'),
    ('A', 'T'), ('T', 'A'),
    ('G', 'C'), ('C', 'G'),
    ('G', 'T'), ('T', 'G')
}
```

```python
contador = 0

posicoes = {}

for base in sequence1:
    
    if base != sequence2[contador]:
        
        if (base, sequence2[contador]) in transicoes:
            posicoes[contador+1] = (base, sequence2[contador], 'transição')
        
        elif (base, sequence2[contador]) in transversoes:
            posicoes[contador+1] = (base, sequence2[contador], 'transversão')

    contador += 1

```
